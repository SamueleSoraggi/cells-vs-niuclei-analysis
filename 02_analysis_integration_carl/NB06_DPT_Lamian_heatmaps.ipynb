```python
import numpy as np
import seaborn as sns
from scipy.interpolate import interp1d

import matplotlib.pyplot as plt

# Get differentially expressed genes per cluster
lamian_diffgene = adata.uns['lamian_diffgene_corr']

# Get unique clusters
clusters = adata.obs['clusters_compact'].unique()

# Set maximum number of genes to display per heatmap
max_genes_display = 100

# Define pseudotime bins for alignment
n_bins = 50
pseudotime_bins = np.linspace(0, 1, n_bins)

for cluster in clusters:
    # Get genes for this cluster
    if cluster not in lamian_diffgene:
        continue
    
    cluster_genes = lamian_diffgene[cluster]
    if len(cluster_genes) == 0:
        continue

    # Filter genes by p-value < 0.001
    if cluster in adata.uns['lamian_stat_corr']:
        cluster_stats = adata.uns['lamian_stat_corr'][cluster]
        # Assuming the stats dictionary has gene names as keys and contains p-values
        print()
        filtered_genes = cluster_stats.iloc[:,0].index.tolist()
        print(filtered_genes)
        if len(filtered_genes) == 0:
            continue
        cluster_genes = filtered_genes
    else:
        continue
    
    # Prepare data for heatmap
    # Limit to max_genes_display
    if len(cluster_genes) > max_genes_display:
        cluster_genes = cluster_genes[:max_genes_display]
    
    # Create matrices for Cells and Nuclei
    cells_matrix = np.zeros((len(cluster_genes), n_bins))
    nuclei_matrix = np.zeros((len(cluster_genes), n_bins))
    
    for i, gene in enumerate(cluster_genes):
        # Get data for Cells batch
        cells_mask = (adata.obs['batch'] == 'Cells') & (adata.obs['clusters_compact'] == cluster)
        cells_data = adata[cells_mask, gene]
        cells_pseudotime = adata.obs.loc[cells_mask, 'dpt_palantir_cellalign'].values
        cells_expr = cells_data.X.toarray().flatten() if hasattr(cells_data.X, 'toarray') else cells_data.X.flatten()
        
        # Sort by pseudotime and bin
        if len(cells_pseudotime) > 0:
            cells_sorted_idx = np.argsort(cells_pseudotime)
            cells_pt_sorted = cells_pseudotime[cells_sorted_idx]
            cells_expr_sorted = cells_expr[cells_sorted_idx]
            
            # Bin and average expression
            for j in range(n_bins):
                bin_start = j / n_bins
                bin_end = (j + 1) / n_bins
                bin_mask = (cells_pt_sorted >= bin_start) & (cells_pt_sorted < bin_end)
                if np.sum(bin_mask) > 0:
                    cells_matrix[i, j] = np.mean(cells_expr_sorted[bin_mask])
        
        # Get data for Nuclei batch
        nuclei_mask = (adata.obs['batch'] == 'Nuclei') & (adata.obs['clusters_compact'] == cluster)
        nuclei_data = adata[nuclei_mask, gene]
        nuclei_pseudotime = adata.obs.loc[nuclei_mask, 'dpt_palantir_cellalign'].values
        nuclei_expr = nuclei_data.X.toarray().flatten() if hasattr(nuclei_data.X, 'toarray') else nuclei_data.X.flatten()
        
        # Sort by pseudotime and bin
        if len(nuclei_pseudotime) > 0:
            nuclei_sorted_idx = np.argsort(nuclei_pseudotime)
            nuclei_pt_sorted = nuclei_pseudotime[nuclei_sorted_idx]
            nuclei_expr_sorted = nuclei_expr[nuclei_sorted_idx]
            
            # Bin and average expression
            for j in range(n_bins):
                bin_start = j / n_bins
                bin_end = (j + 1) / n_bins
                bin_mask = (nuclei_pt_sorted >= bin_start) & (nuclei_pt_sorted < bin_end)
                if np.sum(bin_mask) > 0:
                    nuclei_matrix[i, j] = np.mean(nuclei_expr_sorted[bin_mask])
    
    # Create figure with two heatmaps stacked
    fig, axes = plt.subplots(2, 1, figsize=(12, len(cluster_genes) * 0.25), sharex=True)
    
    # Plot Cells heatmap
    sns.heatmap(cells_matrix, ax=axes[0], cmap='viridis', cbar_kws={'label': 'Expression'},
                yticklabels=cluster_genes, xticklabels=False)
    axes[0].set_title(f'Cluster {cluster} - Cells (p < 0.001)', fontsize=14)
    axes[0].set_ylabel('Genes', fontsize=12)
    
    # Plot Nuclei heatmap
    sns.heatmap(nuclei_matrix, ax=axes[1], cmap='viridis', cbar_kws={'label': 'Expression'},
                yticklabels=cluster_genes, xticklabels=10)
    axes[1].set_title(f'Cluster {cluster} - Nuclei (p < 0.001)', fontsize=14)
    axes[1].set_ylabel('Genes', fontsize=12)
    axes[1].set_xlabel('Pseudotime (0-1)', fontsize=12)
    
    # Set x-axis labels to show pseudotime values
    tick_positions = np.linspace(0, n_bins-1, 11)
    tick_labels = [f'{x:.1f}' for x in np.linspace(0, 1, 11)]
    axes[1].set_xticks(tick_positions)
    axes[1].set_xticklabels(tick_labels)
    
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt

# Get differentially expressed genes per cluster
lamian_diffgene = adata.uns['lamian_diffgene']

# Get unique clusters
clusters = list(adata.uns['lamian_stat'].keys())

# Set maximum number of genes to display per heatmap
max_genes_display = 100

# Define pseudotime bins for alignment
n_bins = 50
pseudotime_bins = np.linspace(0, 1, n_bins)

for cluster in clusters:
    # Get genes for this cluster
    if cluster not in lamian_diffgene:
        continue
    
    cluster_genes = lamian_diffgene[cluster]
    if len(cluster_genes) == 0:
        continue

    # Filter genes by p-value < 0.001
    if cluster in adata.uns['lamian_stat']:
        cluster_stats = adata.uns['lamian_stat'][cluster]
        filtered_genes = cluster_stats.iloc[:,0].index.tolist()
        if len(filtered_genes) == 0:
            continue
        cluster_genes = filtered_genes
    else:
        continue
    
    # Limit to max_genes_display
    if len(cluster_genes) > max_genes_display:
        cluster_genes = cluster_genes[:max_genes_display]
    
    # Create combined matrix: each gene appears twice (Cells first, then Nuclei)
    combined_matrix = np.zeros((len(cluster_genes) * 2, n_bins))
    gene_labels = []
    
    for i, gene in enumerate(cluster_genes):
        # Get data for Cells batch
        cells_mask = (adata.obs['batch'] == 'Cells') & (adata.obs['clusters_single'] == cluster)
        cells_data = adata[cells_mask, gene]
        cells_pseudotime = adata.obs.loc[cells_mask, 'dpt_palantir_cellalign'].values
        cells_expr = cells_data.X.toarray().flatten() if hasattr(cells_data.X, 'toarray') else cells_data.X.flatten()
        
        # Sort by pseudotime and bin
        if len(cells_pseudotime) > 0:
            cells_sorted_idx = np.argsort(cells_pseudotime)
            cells_pt_sorted = cells_pseudotime[cells_sorted_idx]
            cells_expr_sorted = cells_expr[cells_sorted_idx]
            # Add separator line between genes (after cells, before nuclei of same gene)

            # Bin and average expression
            for j in range(n_bins):
                bin_start = j / n_bins
                bin_end = (j + 1) / n_bins
                bin_mask = (cells_pt_sorted >= bin_start) & (cells_pt_sorted < bin_end)
                if np.sum(bin_mask) > 0:
                    combined_matrix[i * 2, j] = np.mean(cells_expr_sorted[bin_mask])
        
        # Get data for Nuclei batch
        nuclei_mask = (adata.obs['batch'] == 'Nuclei') & (adata.obs['clusters_single'] == cluster)
        nuclei_data = adata[nuclei_mask, gene]
        nuclei_pseudotime = adata.obs.loc[nuclei_mask, 'dpt_palantir_cellalign'].values
        nuclei_expr = nuclei_data.X.toarray().flatten() if hasattr(nuclei_data.X, 'toarray') else nuclei_data.X.flatten()
        
        # Sort by pseudotime and bin
        if len(nuclei_pseudotime) > 0:
            nuclei_sorted_idx = np.argsort(nuclei_pseudotime)
            nuclei_pt_sorted = nuclei_pseudotime[nuclei_sorted_idx]
            nuclei_expr_sorted = nuclei_expr[nuclei_sorted_idx]
            
            # Bin and average expression
            for j in range(n_bins):
                bin_start = j / n_bins
                bin_end = (j + 1) / n_bins
                bin_mask = (nuclei_pt_sorted >= bin_start) & (nuclei_pt_sorted < bin_end)
                if np.sum(bin_mask) > 0:
                    combined_matrix[i * 2 + 1, j] = np.mean(nuclei_expr_sorted[bin_mask])
        
        # Add labels for both rows
        gene_labels.append(f'{gene} (Cells)')
        gene_labels.append(f'{gene} (Nuclei)')

        
    
    # Create single heatmap
    fig, ax = plt.subplots(figsize=(12, len(cluster_genes) * 0.5))
    
    sns.heatmap(combined_matrix, ax=ax, cmap='viridis', cbar_kws={'label': 'Expression'},
                yticklabels=gene_labels, xticklabels=10)
    ax.set_title(f'Cluster {cluster} - Cells vs Nuclei (p < 0.001)', fontsize=14)
    ax.set_ylabel('Genes', fontsize=12)
    ax.set_xlabel('Pseudotime (0-1)', fontsize=12)
    
    # Set x-axis labels to show pseudotime values
    tick_positions = np.linspace(0, n_bins-1, 11)
    tick_labels = [f'{x:.1f}' for x in np.linspace(0, 1, 11)]
    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels)
    for i, gene in enumerate(cluster_genes):
        ax.axhline(y=i * 2, color='red', linewidth=0.5, alpha=0.7)
    
        
    
    plt.tight_layout()
    plt.show()
```
